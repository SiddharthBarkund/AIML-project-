```python
# ==============================
# TITANIC DATA PROJECT
# Data Cleaning + Visualization
# ==============================

# 1. Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Style
sns.set(style="whitegrid")

# 2. Load Dataset
df = pd.read_csv("titanic.csv")

print("\n🔹 First 5 Rows:")
print(df.head())

print("\n🔹 Dataset Info:")
print(df.info())

print("\n🔹 Missing Values:")
print(df.isnull().sum())


# ==============================
# 3. DATA CLEANING
# ==============================

# Fill missing Age with median
df['Age'].fillna(df['Age'].median(), inplace=True)

# Drop Cabin column (too many missing values)
if 'Cabin' in df.columns:
    df.drop('Cabin', axis=1, inplace=True)

# Fill Embarked with mode
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)

# Remove duplicates
df.drop_duplicates(inplace=True)

print("\n🔹 Missing Values After Cleaning:")
print(df.isnull().sum())


# ==============================
# 4. FEATURE ENGINEERING
# ==============================

# Family Size
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

# Age Groups
df['AgeGroup'] = pd.cut(df['Age'],
                       bins=[0, 12, 18, 35, 60, 100],
                       labels=['Child', 'Teen', 'Adult', 'MidAge', 'Senior'])

print("\n🔹 Updated Data:")
print(df.head())


# ==============================
# 5. VISUALIZATION
# ==============================

plt.figure(figsize=(18, 12))

# 1. Survival Count
plt.subplot(2, 3, 1)
sns.countplot(x='Survived', data=df)
plt.title("Survival Count")

# 2. Gender vs Survival
plt.subplot(2, 3, 2)
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title("Gender vs Survival")

# 3. Class vs Survival
plt.subplot(2, 3, 3)
sns.countplot(x='Pclass', hue='Survived', data=df)
plt.title("Class vs Survival")

# 4. Fare Distribution
plt.subplot(2, 3, 4)
sns.histplot(df['Fare'], bins=30, kde=True)
plt.title("Fare Distribution")

# 5. Age Distribution
plt.subplot(2, 3, 5)
sns.histplot(df['Age'], bins=30, kde=True)
plt.title("Age Distribution")

# 6. Correlation Heatmap
plt.subplot(2, 3, 6)
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.title("Correlation Heatmap")

plt.tight_layout()
plt.show()


# ==============================
# 6. KEY INSIGHTS (PRINT)
# ==============================

print("\n📊 KEY INSIGHTS:")
print("✔ Females had higher survival rate than males")
print("✔ 1st class passengers survived more")
print("✔ Higher fare → better survival chances")
print("✔ Children had slightly better survival")
print("✔ Family size impacts survival")


# ==============================
# 7. SAVE CLEAN DATA
# ==============================

df.to_csv("cleaned_titanic.csv", index=False)

print("\n✅ Cleaned dataset saved as 'cleaned_titanic.csv'")
```
