```python
import json

def load_data(filename):
    with open(filename, "r", encoding="utf-8-sig") as f:
        return json.load(f)

data = load_data("store_data.json")
print(type(data))
```

```python
def clean_data(data):
    text_to_num = {"one":1, "two":2, "three":3, "four":4, "five":5}
    
    cleaned_data = []
    unique_users = set()

    for user in data["reviews"]:
        
        # rating clean
        raw_rating = str(user["rating"]).strip().lower()
        
        if raw_rating in text_to_num:
            raw_rating = text_to_num[raw_rating]
        else:
            raw_rating = float(raw_rating)
        
        user["rating"] = raw_rating
        
        # handle missing age
        if user.get("age") is None:
            user["age"] = None
        
        # deduplication
        name = user["name"].strip()
        if name in unique_users:
            continue
        
        unique_users.add(user["name"])
        cleaned_data.append(user)

    return cleaned_data
```

```python
data=clean_data(data)
```

```python
def get_insights(data):

    tot_rating = 0
    poor_rating = 0   # ✅ same indentation
    
    for user in data["reviews"]:
        rating = float(user["rating"])
        tot_rating += rating
        
        if rating < 3:
            poor_rating += 1

    avg_rating = tot_rating / len(data["reviews"])

    print("Average Rating:", avg_rating)
    print("Poor Ratings:", poor_rating)
```

```python
get_insights(data)
```

```python
def get_recommendations(data):

    recommendations = []

    for user in data["reviews"]:   # 🔥 important change

        curr_recomm = {}
        curr_recomm["name"] = user["name"]

        if float(user["rating"]) >= 4:
            curr_recomm["brand"] = "apple"
        else:
            curr_recomm["brand"] = "samsung"

        recommendations.append(curr_recomm)

    return recommendations
```

```python
get_recommendations(data)
```
